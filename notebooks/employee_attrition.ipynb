## Gender vs Attrition

```python


# first see in table how many male and female leave and stay

pd.crosstab(
    data['Gender'],
    data['Attrition']
)

# with %
pd.crosstab(
    data['Gender'],
    data['Attrition'],
    normalize='index'
)*100
# we see 14% female and 17% male left
```

```python
# visuzlaize for each gender how many eft and stayed

sns.countplot(data=data,x='Gender',hue='Attrition')
```

```python
# check association between gender and attrition using chi2 value

from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(
    data['Gender'],
    data['Attrition']
)

chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)


# • High Chi2 value + Low P-value (below 0.05) --> Strong Association (There is a real pattern).
# • Low Chi2 value + High P-value (above 0.05) --> No Association (There is no pattern).

```

```python
# the p value is > 0.05 means gender does not determine whether employee will leave or not that strongly
```

```python
# Spearman coefficient = direction and strength.
# +ve and near 1 means as one faeture inc the other also inc i.e +ve monotonic(as one inc other inc dec or same not flcutuate) rlationship
# -ve and away from 1 means as one faeture inc the other dec i.e -ve monotonic(as one inc other inc dec or same not flcutuate) rlationship
#near about 0 value means no clear monotonic relationship --> means as one feature inc the other does not monotonicallly inc or dec or stay same 
# i.e no clear monotonic relationshp can be established
# it doesnt mean theres no relationship between them its just theres no clear monotonic relationship between them


# p-value = statistical evidence that a relationship exists.
# p-value < 0.05: Sufficient statistical evidence to say a relationship exists.
# p-value > 0.05: Insufficient statistical evidence to say a relationship exists.
```

```python
# spearman correlation

from scipy.stats import spearmanr

correlation, p_value_spearman = spearmanr(
    data['DailyRate'],
    data['Attrition'].map({'No': 0, 'Yes': 1})
)

print("Spearman correlation:", correlation)
print("Spearman p-value:", p_value_spearman)
```

```python
# Spearman correlation is negative and very close to zero,
# indicating a very weak negative monotonic relationship
# between DailyRate and Attrition.

# p-value < 0.05 indicates statistically significant
# evidence of a relationship.

# However, the relationship is extremely weak, so
# DailyRate alone has very little monotonic association
# with Attrition.

# a relationship can be statistically significant but weak
```

## multivariate analysis

```python
# analyze relationship between multiple cols by analyzing corr value between them
```

```python
# Correlation measures the direction and strength of the relationship between two numerical variables.
```

```python
# Pearson correlation compares how much two variables vary together (covariance) against how much each variable varies on its own (standard deviation).
# r = underroot (summation ( (xi-xbar) * (yi-ybar) ) ) / underroot ( (summation( (xi-xbar)^2 ) * (summation (yi-ybar)^2 )
# where xi and yi are feature vals in each cell
```

```python
# meausres only strength and direction of linear relationship
# if relp is strong but not linear then pearson corr is close to 0
```

```python
# extract numerical cols

num_cols_df=data.select_dtypes(['int64','float64'])

num_cols_df
```

```python
# convert op attrition(categorical) to numerical and store in num_cols_df as col

num_cols_df['Attrition']=data['Attrition'].map({
    'No':0,
    'Yes':1
})

num_cols_df['Attrition'].value_counts()
```

```python
# correlation matrix

corr_matrix=num_cols_df.corr(method='pearson')

corr_matrix
```

```python
# see in the form of a heatmap

sns.heatmap(corr_matrix,cmap='coolwarm',vmin=-1,vmax=1)
```

```python
# reddish zones high +ve correlation and dark blue zones have high -ve correlation

# this mask will be applied on each sq of corr matrix 
strong_corr=np.abs(corr_matrix) < 0.7

# np.abs means take abs value for +ve stays same for -ve the -ve sign is removed i.e np.abs(-0.8) becomes 0.8
# now this mask sets values < 0.7 as true and these have to be hidden
# values > 0.7 are false and have to be shown

sns.heatmap(corr_matrix,cmap='coolwarm',mask=strong_corr,vmin=-1,vmax=1)
```

```python
# remove the repeated scores that occur in upper traingle and the diagonals that depict relationship of cols with themselves

# 1. Hide weak correlations (|r| < 0.7)
weak_mask = np.abs(corr_matrix) < 0.7

# 2. Hide the upper triangle AND the main diagonal (k=0 includes the diagonal)
upper_diag_mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=0)

# 3. Combine both masks using the OR operator (|)
final_mask = weak_mask | upper_diag_mask

plt.figure(figsize=(10,5))
sns.heatmap(
    corr_matrix,
    mask=final_mask,
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    annot=True,
    fmt='.2f'
)
plt.title('Unique Strong Correlations (|r| >= 0.7)')
plt.show()
```

```python

# JobLevel – MonthlyIncome	0.95	Very strong relationship. Higher job levels are associated with higher monthly income.
# JobLevel – TotalWorkingYears	0.78	Employees with more total work experience tend to have higher job levels.
# MonthlyIncome – TotalWorkingYears	0.77	More experienced employees tend to earn more.
# PercentSalaryHike – PerformanceRating	0.77	Higher performance ratings are associated with higher salary hikes.
# YearsAtCompany – YearsInCurrentRole	0.76	Employees who have spent longer at the company tend to have spent longer in their current role.
# YearsAtCompany – YearsSinceLastPromotion	0.77	Longer company tenure is associated with more time since the last promotion.
# YearsInCurrentRole – YearsSinceLastPromotion	0.71	Longer time in the current role tends to be associated with more time since the last promotion.
```

```python
# higher correlation like 0.95 means multicollinearity exists
# these cols carry overlapping info
# linear models like logistic reg wont be able to assign weights properly if bothof these are included 
# for tree based models like randomforest xbgoost this does ont matter much 
```

### VIF(Variance Inflation Factor)

```python
# ==============================================================================
# Variance Inflation Factor (VIF) & Multicollinearity Notes
# ==============================================================================
# 1. R-squared (R2): Measures how much variance (information) in one column 
#    can be predicted using all the other columns combined.
#    - Example: R2 = 0.90 means 90% of that feature's info is already captured 
#      by other features.
#
# 2. VIF Formula: VIF = 1 / (1 - R2)
#    - As R2 increases, the denominator (1 - R2) shrinks, causing VIF to spike.
#
# 3. Decision Thresholds:
#    - VIF = 1      -> No multicollinearity (completely independent feature)
#    - VIF 1 to 5   -> Low to moderate correlation (generally safe to keep)
#    - VIF > 5      -> High multicollinearity (consider dropping or combining)
#    - VIF > 10     -> Severe multicollinearity (drop redundant column)
# ==============================================================================
```

```python
# VIF is based on how well a feature can be explained through a linear combination of the other features. 
    # It measures linear multicollinearity, not every possible nonlinear relationship.
```

```python
# Select all numerical columns
# Remove Attrition because it is our target variable
# Remove EmployeeNumber because it is only an employee identifier

X = data.select_dtypes(include='number').drop(
    columns=['Attrition', 'EmployeeNumber']
)
```

```python
from statsmodels.tools.tools import add_constant

# Add a constant column (intercept) for Linear Regression
# VIF calculations should account for the intercept

X = add_constant(X)
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_data = pd.DataFrame()

# Store feature names (excluding the constant column)
vif_data['Feature'] = X.columns[1:]

# Calculate VIF for every independent feature
vif_data['VIF'] = [
    variance_inflation_factor(X.values, i)
    for i in range(1, X.shape[1])
]

# Display features in descending order of VIF
vif_data.sort_values(by='VIF', ascending=False)
```

```python


# A high VIF indicates that a feature shares considerable information with other independent features. This can make regression coefficients unstable and difficult to interpret.

# **Important considerations:**
# - Attrition is excluded because it is the target variable.
# - EmployeeNumber is excluded because it is an identifier, not a meaningful predictor.
# - A constant is added to account for the intercept in regression.
# - Numerical representations of ordinal categorical features are included in this initial analysis, but their coding assumptions must be considered when interpreting VIF.
# - String-based categorical features are not included in this initial numerical VIF calculation. They will require appropriate encoding for a complete model-level analysis.

# High VIF alone does not mean a feature must be removed. Feature removal should depend on domain knowledge, model requirements and predictive performance.
```

## CROSS VALIDATION

```python
# we have manually selected paramaters for all
# from above table xgboost and lr(threshold=0.3) appear to be close in terms f1 score though xgboost has best recall
# we will use k folds validation to select the best hyperparameters and do this now on training data by splitting it into k parts
# train on k-1 parts then test on 1 part repeat this for all k parts then take average of predicttions as the final result
```

```python
# we use StratifiedKFold to keep the proportion of attritinion(1) and non attritioon(0) equivalent in each fold

from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(
    n_splits=5,  # divide training data into 5 parts 
    shuffle=True, # every time shuffle the rows between them
    random_state=42  # generate the same results
)
```

```python
from sklearn.model_selection import cross_validate
from xgboost import XGBClassifier

xgb_cv_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=2,
    min_child_weight=7,
    subsample=1.0,
    colsample_bytree=1.0,
    reg_lambda=1,
    scale_pos_weight=scale_weight,
    random_state=42,
    eval_metric='logloss'
)

scoring = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1',
    'roc_auc': 'roc_auc',
    'pr_auc': 'average_precision'
}

cv_results = cross_validate(
    xgb_cv_model, # model being used and its hyperparameters
    X_train,  # ip set for training
    y_train,  # op result of training
    cv=skf,   # cross validate using stratified k fold
    scoring=scoring
)

for metric in scoring:
    scores = cv_results[f'test_{metric}']
    print(f"{metric}: {scores.mean():.3f}")
```

```python
### SHAP Summary Plot Analysis: Key Drivers of Employee Attrition

# **Overview:**
# The SHAP beeswarm plot ranks features from top to bottom based on their overall impact on the model's predictions. Each dot represents a single employee:
# * **Position on X-axis (SHAP Value):** Dots to the right (`> 0`) increase the predicted probability of attrition; dots to the left (`< 0`) decrease it.
# * **Color (Feature Value):** **Red** indicates a high feature value (or `1` for binary variables), while **blue** indicates a low feature value (or `0`).

# **Feature-by-Feature Conclusions:**
# 1. **OverTime (`OverTime_Yes`):** The strongest driver of attrition. Employees working overtime (red, `1`) experience a sharp positive push in attrition risk, whereas employees without overtime (blue, `0`) consistently have lower predicted turnover.
# 2. **Stock Option Level (`StockOptionLevel`):** Employees with zero stock options (blue, `0`) show a higher risk of leaving. Providing stock options (levels `1–3`, purple/red) strongly reduces attrition risk.
# 3. **Age (`Age`):** Younger employees (blue) are significantly more likely to leave, while older, more senior employees (red) tend to stay with the company.
# 4. **Monthly Income (`MonthlyIncome`):** Lower salaries (blue) push predictions toward attrition, whereas higher compensation (red) acts as a strong retention factor.
# 5. **Number of Companies Worked (`NumCompaniesWorked`):** Employees who have worked at many previous companies (red) show a higher likelihood of leaving compared to those with fewer past employers (blue).
# 6. **Years With Current Manager (`YearsWithCurrManager`):** Employees who are newly assigned to a manager (`0–1` years, blue) face higher attrition risk, while long-term manager stability (red/purple) lowers turnover.
# 7. **Environment Satisfaction (`EnvironmentSatisfaction`):** Low workplace satisfaction scores (blue, rating `1`) strongly increase attrition risk, while high satisfaction (red, ratings `3–4`) helps retain employees.
# 8. **Distance From Home (`DistanceFromHome`):** Longer daily commutes (red) modestly increase the likelihood of attrition compared to shorter commutes (blue).
# 9. **R&D Department (`Department_Research & Development`):** Employees working in Research & Development (red, `1`) have a slightly lower predicted risk of leaving than employees in other departments such as Sales or HR (blue, `0`).
# 10. **Remaining Features (`Sum of 34 other features`):** While individual secondary features have smaller effects, their combined contribution can still meaningfully shift an employee's final risk score.

# **Business Recommendations:**
# * **Control Workload:** Monitor and limit excessive overtime, especially among younger and lower-paid staff.
# * **Early Retention Focus:** Support employees during their first year under a new manager and review compensation or stock option packages for high-risk roles.
```

```python
### SHAP Waterfall Plot Analysis: Individual Employee Prediction

# **How to Read This Plot (Bottom to Top):**
# The SHAP waterfall plot explains how the model arrived at the exact prediction for **one single employee**. Read it like a step-by-step math equation starting at the bottom and moving up:
# * Starting Baseline (Bottom): E[f(X)] = 0.002 is the model's average starting guess across all employees before looking at this person's details.
# * Grey Numbers (Left): The actual data values recorded for this specific employee (for example, 2033 = MonthlyIncome means they earn $2,033/month).
# * Red Arrows (+): Positive impact. Taking this feature into account pushes the prediction UP toward attrition (higher risk of leaving).
# * Blue Arrows (-): Negative impact. Taking this feature into account pushes the prediction DOWN toward staying (lower risk of leaving).
# * Final Score (Top): f(x) = 0.961 is the final combined score after adding and subtracting every feature. Because 0.961 is well above 0, the model predicts this employee is at high risk of leaving.

# **Feature-by-Feature Breakdown (From Bottom to Top):**
# 1. 34 other features (impact: -0.14): The combined effect of the 34 smaller features slightly pushes the prediction toward staying (moves score from 0.002 to -0.138).
# 2. JobRole_Sales Representative = 1 (impact: +0.16): Working as a Sales Representative pushes the prediction slightly toward attrition (moves score to +0.022).
# 3. Department_Research & Development = 0 (impact: +0.25): Not working in the R&D department pushes the prediction toward attrition (moves score to +0.272).
# 4. StockOptionLevel = 1 (impact: -0.26): Having level 1 stock options acts as a retention factor and pushes the prediction toward staying (moves score back down to +0.012).
# 5. Age = 24 (impact: +0.29): Being 24 years old pushes the prediction toward attrition, as younger employees have higher turnover rates (moves score to +0.302).
# 6. TotalWorkingYears = 1 (impact: +0.36): Having only 1 year of total work experience strongly pushes the prediction toward attrition (moves score to +0.662).
# 7. BusinessTravel = 0 (impact: -0.44): Not having to travel for business significantly lowers risk and pushes the prediction toward staying (moves score down to +0.222).
# 8. OverTime_Yes = 0 (impact: -0.50): Not working overtime is the strongest protective factor here, pushing the prediction strongly toward staying (moves score down to -0.278).
# 9. YearsWithCurrManager = 0 (impact: +0.58): Having 0 years with their current manager creates a major jump toward attrition (moves score up to +0.302).
# 10. MonthlyIncome = 2033 (impact: +0.66): Earning a low monthly salary of $2,033 is the single largest risk driver for this employee, pushing the final score all the way up to 0.961.

# **Conclusion:**
# Even though this employee has positive working conditions—no overtime (-0.50), no business travel (-0.44), and level 1 stock options (-0.26)—those protective factors are outweighed by their low monthly salary ($2,033, +0.66), being brand new to their manager (0 years, +0.58), and early career stage (1 working year, +0.36 and age 24, +0.29). Together, these factors drive the final prediction to f(x) = 0.961 (high risk of attrition). To retain this employee, HR should prioritize a compensation review and closer manager onboarding support.
```

```python
# • The Impact Value (SHAP Value): It is the exact amount by which a feature changed the model's prediction once that feature was taken into account.
# • Positive Impact (+): Pushes the prediction toward attrition (increases the predicted risk of leaving).
# • Negative Impact (-): Pushes the prediction toward staying (decreases the predicted risk of leaving).
# • Zero Impact (0): Has no effect on the prediction (the model's guess stays the same whether it sees that feature or not).
```
